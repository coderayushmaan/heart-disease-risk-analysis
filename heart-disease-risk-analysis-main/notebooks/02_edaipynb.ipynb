```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#load dataset
df = pd.read_csv("heart.csv")
```

```python
#check for null values
print("NULL VALUES:")
print(df.isnull().sum())
```

```python
#duplicate values
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
#Summary Stastistics for outlier Detection
print("\nSummary Statistics:")
print(df.describe())
```

```python
#Distribution Plot Age
plt.figure(figsize=(6,4))
sns.histplot(df['age'], kde = True)
plt.title('Age Distribution')
plt.show()
```

```python
#Distribution Plot Cholesterol
plt.figure(figsize=(6,4))
sns.histplot(df['chol'], kde = True)
plt.title('Cholesterol Distribution')
plt.show()
```

```python
#Count Plot Of Target
plt.figure(figsize=(6,4))
sns.countplot(x='target', data=df)
plt.title('Target CLass Distribution')
plt.xlabel('Target (0 = No Disease, 1 = Disease)')
plt.ylabel('Count')
plt.show()
```

```python
#BoxPlot Comparing Numeric Features by Target Class
numeric_features = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']
for col in numeric_features:
    plt.figure(figsize=(6,4))
    sns.boxplot(x='target', y=col, data=df)
    plt.title(f'{col} by Target Class')
    plt.show()
```

```python
#Bar Chart Chest Pain Type
plt.figure(figsize=(6,4))
sns.countplot(x='cp', data=df)
plt.title('Chest Pain Type Distribution')
plt.xlabel('Chest Pain Type')
plt.ylabel('Count')
plt.show()
```

```python
#Bar Chart : Sex
plt.figure(figsize=(6,4))
sns.countplot(x='sex', data=df)
plt.title('Sex Distribution')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.show()
```

```python
# Bar Chart: Fasting Blood Sugar
plt.figure(figsize=(6,4))
sns.countplot(x='fbs', data=df)
plt.title('Fasting Blood Sugar Distribution')
plt.xlabel('FBS')
plt.ylabel('Count')
plt.show()
```
