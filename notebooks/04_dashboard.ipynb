```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("heart.csv")
```

```python
df.head()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(20, 12))
fig.suptitle("Heart Disease Dashboard", fontsize=20)

sns.countplot(data=df, x='target', ax=axes[0,0])
sns.histplot(data=df, x='age', hue='target', kde=True, ax=axes[0,1])
sns.countplot(data=df, x='cp', ax=axes[0,2])
sns.heatmap(df.corr(), cmap='coolwarm', ax=axes[1,0])
sns.scatterplot(data=df, x='age', y='chol', hue='target', ax=axes[1,1])
sns.boxplot(data=df, x='sex', y='thalach', ax=axes[1,2])

plt.tight_layout(rect=[0,0,1,0.96])
plt.savefig("heart_disease_dashboard.png", dpi=300, bbox_inches="tight")
plt.show()
```
