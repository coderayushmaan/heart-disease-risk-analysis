```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import chi2_contingency
```

```python
df = pd.read_csv("heart.csv")
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Pearson Correlation Heatmap')
plt.show()
```

```python
print(df.corr()['target'].sort_values(ascending=False))
```

```python
# Scatter Plot: Age vs Maximum Heart Rate
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x='age', y='thalach', hue='target')
plt.title('Age vs Maximum Heart Rate')
plt.show()
```

```python
# Chi-Square Test: Sex vs Target
contingency_sex = pd.crosstab(df['sex'], df['target'])
chi2, p, dof, expected = chi2_contingency(contingency_sex)

print("Sex vs Target")
print("Chi-square Statistic:", chi2)
print("P-value:", p)
```

```python

# Chi-Square Test: Chest Pain Type vs Target
contingency_cp = pd.crosstab(df['cp'], df['target'])
chi2, p, dof, expected = chi2_contingency(contingency_cp)

print("\nChest Pain Type vs Target")
print("Chi-square Statistic:", chi2)
print("P-value:", p)
```

```python
# Chi-Square Test: Fasting Blood Sugar vs Target
contingency_fbs = pd.crosstab(df['fbs'], df['target'])
chi2, p, dof, expected = chi2_contingency(contingency_fbs)

print("\nFasting Blood Sugar vs Target")
print("Chi-square Statistic:", chi2)
print("P-value:", p)
```
